```python
import os
import textwrap
import torch

# 1. Set Working Directory
%cd /content/SepReformer

# 2. Reset Files to Clean State
# This prevents patch conflicts if the cell is run multiple times.
print("🔄 Resetting code files...")
!git checkout models/SepReformer_Base_WSJ0/modules/module.py
!git checkout models/SepReformer_Base_WSJ0/model.py
!git checkout models/SepReformer_Base_WSJ0/engine.py
!git checkout models/SepReformer_Base_WSJ0/dataset.py

# 3. Write Configuration
print("⚙️ Writing config file...")
config_content = """
project: "[Project] SepReformer"
notes: "Training on custom dataset - Fixed Scheduler"

config:
    dataset:
        max_len: 32000
        sampling_rate: 8000
        scp_dir: "data/custom"
        train:
            mixture: "train_mix_clean.scp"
            spk1: "train_s1.scp"
            spk2: "train_s2.scp"
            dynamic_mixing: true
        valid:
            mixture: "val_mix_clean.scp"
            spk1: "val_s1.scp"
            spk2: "val_s2.scp"
        test:
            mixture: "val_mix_clean.scp"
            spk1: "val_s1.scp"
            spk2: "val_s2.scp"

    dataloader:
        batch_size: 1
        pin_memory: true
        num_workers: 4
        drop_last: true

    model:
        num_stages: 4
        num_spks: 2
        module_audio_enc:
            in_channels: 1
            out_channels: 256
            kernel_size: 16
            stride: 4
            groups: 1
            bias: false
        module_feature_projector:
            num_channels: 256
            in_channels: 256
            out_channels: 128
            kernel_size: 1
            bias: false
        module_separator:
            num_stages: 4
            relative_positional_encoding:
                in_channels: 128
                num_heads: 8
                maxlen: 2000
                embed_v: false
            enc_stage:
                global_blocks:
                    in_channels: 128
                    num_mha_heads: 8
                    dropout_rate: 0.05
                local_blocks:
                    in_channels: 128
                    kernel_size: 65
                    dropout_rate: 0.05
                down_conv_layer:
                    in_channels: 128
                    samp_kernel_size: 5
            spk_split_stage:
                in_channels: 128
                num_spks: 2
            simple_fusion:
                out_channels: 128
            dec_stage:
                num_spks: 2
                global_blocks:
                    in_channels: 128
                    num_mha_heads: 8
                    dropout_rate: 0.05
                local_blocks:
                    in_channels: 128
                    kernel_size: 65
                    dropout_rate: 0.05
                spk_attention:
                    in_channels: 128
                    num_mha_heads: 8
                    dropout_rate: 0.05
        module_output_layer:
            in_channels: 256
            out_channels: 128
            num_spks: 2
        module_audio_dec:
            in_channels: 256
            out_channels: 1
            kernel_size: 16
            stride: 4
            bias: false

    criterion:
        name: ["PIT_SISNR_mag", "PIT_SISNR_time", "PIT_SISNRi", "PIT_SDRi"]
        PIT_SISNR_mag:
            frame_length: 512
            frame_shift: 128
            window: 'hann'
            num_stages: 4
            num_spks: 2
            scale_inv: true
            mel_opt: false
        PIT_SISNR_time:
            num_spks: 2
            scale_inv: true
        PIT_SISNRi:
            num_spks: 2
            scale_inv: true
        PIT_SDRi:
            dump: 0

    optimizer:
        name: ["AdamW"]
        AdamW:
            lr: 1.0e-3
            weight_decay: 1.0e-2

    scheduler:
        name: ["ReduceLROnPlateau", "WarmupConstantSchedule"]
        ReduceLROnPlateau:
            mode: "min"
            min_lr: 1.0e-10
            factor: 0.8
            patience: 2
        WarmupConstantSchedule:
            warmup_steps: 100

    check_computations:
        dummy_len: 16000

    engine:
        max_epoch: 90
        gpuid: "0"
        mvn: false
        clip_norm: 5
        start_scheduling: 10
        test_epochs: [50, 90]
"""
with open("models/SepReformer_Base_WSJ0/configs.yaml", "w") as f:
    f.write(config_content)
!rm -rf models/SepReformer_Base_WSJ0/log

# 4. Patch dataset.py
print("🔧 Patching dataset.py...")
dataset_path = 'models/SepReformer_Base_WSJ0/dataset.py'
with open(dataset_path, 'r') as f:
    lines = f.readlines()

new_lines = []
for line in lines:
    if "key.split('_')" in line and "key_random.split('_')" in line:
        indent = line[:line.find("tmp1")]
        new_lines.append(f"{indent}tmp1 = True # Patched\n")
    elif "tmp2 =" in line and "key.split" in line:
        indent = line[:line.find("tmp2")]
        new_lines.append(f"{indent}tmp2 = True # Patched\n")
    elif 'self.speed_aug(torch.tensor(samps_tmp))' in line:
        new_lines.append(f"            # {line.strip()} # PATCHED: Disabled\n")
    else:
        new_lines.append(line)

with open(dataset_path, 'w') as f:
    f.writelines(new_lines)

# 5. Patch engine.py
print("🔧 Patching engine.py...")
engine_path = 'models/SepReformer_Base_WSJ0/engine.py'
with open(engine_path, 'r') as f:
    lines = f.readlines()

new_lines = []
for line in lines:
    if 'util_engine.save_checkpoint_per_best' in line:
        indent = line.split("valid_loss_best")[0]
        new_line = (
            f"{indent}valid_loss_best = util_engine.save_checkpoint_per_best("
            f"valid_loss_best, valid_loss_src_time, train_loss_src_time, epoch, "
            f"self.model, self.main_optimizer, self.checkpoint_path)\n"
        )
        new_lines.append(new_line)
    elif 'writer_src.add_scalars("Learning Rate"' in line:
        new_lines.append(line.replace('add_scalars', 'add_scalar'))
    else:
        new_lines.append(line)

with open(engine_path, 'w') as f:
    f.writelines(new_lines)

# 6. Patch module.py & model.py (AttentionSpeakerSplit)
print("🔧 Patching model architecture (AttentionSpeakerSplit)...")
module_path = 'models/SepReformer_Base_WSJ0/modules/module.py'
model_path = 'models/SepReformer_Base_WSJ0/model.py'

NEW_CLASS_CODE = textwrap.dedent("""
# ─────────────────────────────────────────────────────────────────────────────
#  AttentionSpeakerSplit  (drop-in replacement for SpkSplitStage)
# ─────────────────────────────────────────────────────────────────────────────
class AttentionSpeakerSplit(torch.nn.Module):
    def __init__(self, in_channels: int, num_spks: int,
                 num_heads: int = 8, dropout: float = 0.05):
        super().__init__()
        self.in_channels = in_channels
        self.num_spks    = num_spks
        self.speaker_queries = torch.nn.Parameter(torch.empty(num_spks, in_channels))
        torch.nn.init.trunc_normal_(self.speaker_queries, std=0.02)
        self.cross_attn = torch.nn.MultiheadAttention(
            embed_dim=in_channels, num_heads=num_heads, dropout=dropout, batch_first=False
        )
        self.attn_norm = torch.nn.LayerNorm(in_channels)
        self.ff_gate = torch.nn.Sequential(
            torch.nn.LayerNorm(in_channels),
            torch.nn.Linear(in_channels, in_channels * 2, bias=True),
            torch.nn.GLU(dim=-1),
        )
        self.ff_norm = torch.nn.LayerNorm(in_channels)
        self.res_proj = torch.nn.Linear(in_channels, in_channels, bias=False)
        self.out_norm = torch.nn.GroupNorm(1, in_channels)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, D, T = x.shape
        kv = x.permute(2, 0, 1)
        q  = self.speaker_queries.unsqueeze(1).expand(-1, B, -1)

        # PTFLOPS FIX: Pass q, k, v as positional arguments.
        attn_out, _ = self.cross_attn(q, kv, kv, need_weights=False)

        attn_out = self.attn_norm(attn_out + q)
        x_exp   = x.unsqueeze(1).expand(-1, self.num_spks, -1, -1)
        spk_ctx = attn_out.permute(1, 0, 2).unsqueeze(-1)
        res = self.res_proj(x.permute(0, 2, 1))
        res = res.permute(0, 2, 1).unsqueeze(1)
        out = x_exp * torch.sigmoid(spk_ctx) + res
        out_flat = out.reshape(B * self.num_spks, D, T).permute(0, 2, 1)
        ff_out   = self.ff_gate(out_flat)
        out_flat = self.ff_norm(out_flat + ff_out)
        out      = out_flat.permute(0, 2, 1).reshape(B, self.num_spks, D, T)

        # Reshape to (B*S, D, T) for GroupNorm and final output
        out = self.out_norm(out.reshape(B * self.num_spks, D, T))
        return out
""")

# Patch module.py
with open(module_path, 'r') as f:
    src = f.read()
src = src.replace("class SpkSplitStage", "class _OldSpkSplitStage")
src = src.replace("SpkSplitStage(", "AttentionSpeakerSplit(")
src += NEW_CLASS_CODE
with open(module_path, 'w') as f:
    f.write(src)

# Patch model.py
with open(model_path, 'r') as f:
    src = f.read()
src = src.replace("SpkSplitStage(", "AttentionSpeakerSplit(")
with open(model_path, 'w') as f:
    f.write(src)

# 1. Repair dataset.py (Fix Syntax Error)
print("🔧 Repairing dataset.py...")
# Reset to clean state first
!git checkout models/SepReformer_Base_WSJ0/dataset.py

dataset_path = 'models/SepReformer_Base_WSJ0/dataset.py'
with open(dataset_path, 'r') as f:
    lines = f.readlines()

new_lines = []
for line in lines:
    # Explicitly check for tmp1/tmp2 assignment to avoid partial matching bugs
    if "tmp1 =" in line and "key.split" in line:
        indent = line[:line.find("tmp1")]
        new_lines.append(f"{indent}tmp1 = True # Patched\n")
    elif "tmp2 =" in line and "key.split" in line:
        indent = line[:line.find("tmp2")]
        new_lines.append(f"{indent}tmp2 = True # Patched\n")
    elif 'self.speed_aug(torch.tensor(samps_tmp))' in line:
        new_lines.append(f"            # {line.strip()} # PATCHED: Disabled\n")
    else:
        new_lines.append(line)

with open(dataset_path, 'w') as f:
    f.writelines(new_lines)
print("✅ dataset.py repaired and patched.")



print("🎉 Setup Complete: Config updated and all patches applied!")
```

```python
import os
import yaml
import random

# --- Speed Up Strategy: Train on a Subset (Mini-Epochs) ---
# Instead of iterating all ~12k files, we pick a random subset.
# This makes each "epoch" much faster, allowing frequent validation/saving.
subset_size = 1000  # <--- Adjust this to control epoch length
scp_dir = 'data/custom'
config_path = 'models/SepReformer_Base_WSJ0/configs.yaml'

print(f"✂️  Preparing mini-dataset ({subset_size} samples) for faster iteration...")

if os.path.exists(scp_dir) and os.path.exists(config_path):
    # 1. Define source files (The Full Dataset)
    files_map = {
        'mixture': 'train_mix_clean.scp',
        'spk1': 'train_s1.scp',
        'spk2': 'train_s2.scp'
    }

    try:
        # Read original full datasets
        lines_data = {}
        for key, fname in files_map.items():
            with open(os.path.join(scp_dir, fname), 'r') as f:
                lines_data[key] = f.readlines()

        num_samples = len(lines_data['mixture'])

        if num_samples > subset_size:
            # 2. Randomly select indices
            indices = sorted(random.sample(range(num_samples), subset_size))

            # 3. Write mini scp files
            new_files_map = {}
            for key, fname in files_map.items():
                # Extract subset
                subset_lines = [lines_data[key][i] for i in indices]

                # New filename
                new_fname = fname.replace("train_", "train_mini_")
                new_path = os.path.join(scp_dir, new_fname)

                with open(new_path, 'w') as f:
                    f.writelines(subset_lines)
                new_files_map[key] = new_fname

            print(f"   Created subset files in {scp_dir} ({subset_size} samples)")

            # 4. Update Config
            with open(config_path, 'r') as f:
                config = yaml.safe_load(f)

            # Update pointers to the mini dataset
            if 'config' in config and 'dataset' in config['config']:
                config['config']['dataset']['train']['mixture'] = new_files_map['mixture']
                config['config']['dataset']['train']['spk1'] = new_files_map['spk1']
                config['config']['dataset']['train']['spk2'] = new_files_map['spk2']

                with open(config_path, 'w') as f:
                    yaml.safe_dump(config, f)
                print("✅ Configuration updated to use mini-dataset.")
        else:
            print(f"ℹ️  Dataset size ({num_samples}) is already <= subset_size. Using full dataset.")

    except Exception as e:
        print(f"⚠️  Error creating subset: {e}. Proceeding with existing config.")
else:
    print("⚠️  Data or Config not found. Proceeding...")

print("\n🚀 Starting Training...")
!TF_CPP_MIN_LOG_LEVEL=3 LOGURU_LEVEL=WARNING PYTORCH_ALLOC_CONF=expandable_segments:True python run.py --model SepReformer_Base_WSJ0 --engine-mode train
```

```python
import os
import glob
import random
import shutil
import IPython.display as ipd
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
import mir_eval # Import for metrics

# 1. Define Paths
checkpoint_path = "/content/SepReformer/models/SepReformer_Base_WSJ0/log/scratch_weights/"
# Find the latest checkpoint automatically
checkpoints = sorted(glob.glob(os.path.join(checkpoint_path, "epoch.*.pth")))
latest_ckpt = checkpoints[-1]
print(f"⭐ Using Checkpoint: {os.path.basename(latest_ckpt)}")

# 2. Pick a random test file
# Changed to local dataset path
mix_dir = '/content/dataset/mix_clean'
files = [f for f in os.listdir(mix_dir) if f.endswith('.wav') and '_in' not in f and '_out' not in f]
test_file = random.choice(files)
test_file_path = os.path.join(mix_dir, test_file)

print(f"🎵 Separating: {test_file}")

# 3. Run Inference
!python run.py \
  --model SepReformer_Base_WSJ0 \
  --engine-mode infer_sample \
  --sample-file "{test_file_path}" \
  --out-wav-dir "final_results_90ep" > /dev/null 2>&1

# 4. Locate and Move Files (Handling the path bug we saw earlier)
# The model likely saved them in the source folder again due to the bug
base_name = os.path.splitext(test_file)[0]
path_mix = os.path.join(mix_dir, f"{base_name}_in.wav")
path_s1 = os.path.join(mix_dir, f"{base_name}_out_0.wav")
path_s2 = os.path.join(mix_dir, f"{base_name}_out_1.wav")

if os.path.exists(path_s1):
    print("✅ Separation successful!\n")

    # Audio Player
    print("🎧 Original Mixture:")
    ipd.display(ipd.Audio(path_mix))

    print("🗣️ Speaker 1 (Est):")
    ipd.display(ipd.Audio(path_s1))

    print("🗣️ Speaker 2 (Est):")
    ipd.display(ipd.Audio(path_s2))

    # Visualization
    fig, ax = plt.subplots(1, 3, figsize=(18, 5))

    def plot_spec(ax, p, title):
        y, sr = librosa.load(p, sr=8000)
        D = librosa.amplitude_to_db(np.abs(librosa.stft(y)), ref=np.max)
        img = librosa.display.specshow(D, y_axis='linear', x_axis='time', sr=sr, ax=ax)
        ax.set_title(title)
        return img, y

    img, y_mix = plot_spec(ax[0], path_mix, "Mixture")
    _, y_s1_est = plot_spec(ax[1], path_s1, "Speaker 1 (Est)")
    _, y_s2_est = plot_spec(ax[2], path_s2, "Speaker 2 (Est)")
    fig.colorbar(img, ax=ax, format="%+2.0f dB")
    plt.show()

    # --- 5. Technical Metrics (Ground Truth Comparison) ---
    print("\n📊 Technical Separation Metrics:")

    # Construct GT paths (Assuming s1/s2 sibling folders)
    base_speech_dir = os.path.dirname(mix_dir)
    gt_s1_path = os.path.join(base_speech_dir, 's1', test_file)
    gt_s2_path = os.path.join(base_speech_dir, 's2', test_file)

    if os.path.exists(gt_s1_path) and os.path.exists(gt_s2_path):
        # Load GT
        y_s1_gt, _ = librosa.load(gt_s1_path, sr=8000)
        y_s2_gt, _ = librosa.load(gt_s2_path, sr=8000)

        # Truncate to min length
        min_L = min(len(y_mix), len(y_s1_gt), len(y_s2_gt), len(y_s1_est), len(y_s2_est))
        ref = np.vstack([y_s1_gt[:min_L], y_s2_gt[:min_L]])
        est = np.vstack([y_s1_est[:min_L], y_s2_est[:min_L]])
        mix_base = np.vstack([y_mix[:min_L], y_mix[:min_L]])

        # Calculate BSS Eval (SDR, SIR, SAR)
        # compute_permutation=True allows mir_eval to match speakers automatically
        sdr, sir, sar, _ = mir_eval.separation.bss_eval_sources(ref, est, compute_permutation=True)
        sdr_base, _, _, _ = mir_eval.separation.bss_eval_sources(ref, mix_base, compute_permutation=False) # No perm for mixture baseline

        # Calculate SI-SNR
        def si_snr(target, estimate):
            # Zero mean
            target = target - np.mean(target)
            estimate = estimate - np.mean(estimate)
            dot = np.sum(target * estimate)
            s_target = (dot * target) / np.sum(target ** 2)
            e_noise = estimate - s_target
            return 10 * np.log10(np.sum(s_target ** 2) / np.sum(e_noise ** 2))

        # We assume mir_eval resolved permutation for SDR, but we need it for SI-SNR.
        # Simpler approach: Calculate all permutations for SI-SNR and take max.
        sisnr_1 = si_snr(ref[0], est[0]) + si_snr(ref[1], est[1])
        sisnr_2 = si_snr(ref[0], est[1]) + si_snr(ref[1], est[0])
        avg_sisnr = max(sisnr_1, sisnr_2) / 2

        # Baseline SI-SNR
        avg_base_sisnr = (si_snr(ref[0], mix_base[0]) + si_snr(ref[1], mix_base[1])) / 2

        print(f"  • SI-SNR Improvement (SI-SNRi): {avg_sisnr - avg_base_sisnr:.2f} dB")
        print(f"  • SDR Improvement (SDRi):       {np.mean(sdr - sdr_base):.2f} dB")
        print(f"  • SIR (Interference Rejection): {np.mean(sir):.2f} dB")
        print(f"  • SAR (Artifacts):              {np.mean(sar):.2f} dB")
    else:
        print(f"  ⚠️ Ground truth files not found at {gt_s1_path}. Cannot calculate metrics.")

else:
    print("❌ Output files not found. Check if the model saved to a different location.")
```

```python
import os
import glob
import matplotlib.pyplot as plt
from tensorboard.backend.event_processing.event_accumulator import EventAccumulator

# 1. Define Root Search Directory
root_dir = "models/SepReformer_Base_WSJ0"

# 2. Find all event files recursively
print(f"🔍 Searching for event files in {root_dir}...")
event_files = []
for root, dirs, files in os.walk(root_dir):
    for file in files:
        if "events.out.tfevents" in file:
            event_files.append(os.path.join(root, file))

event_files = sorted(event_files)

if not event_files:
    print(f"❌ No TensorBoard event files found in {root_dir} or its subdirectories.")
    print("ℹ️ Note: If you restored the model from Drive, the training logs (events) might not have been saved/restored, only the checkpoints were.")
else:
    print(f"✅ Found {len(event_files)} event file(s):")
    for f in event_files:
        print(f"  - {f}")

    print("\nParsing events... (this may take a moment)")

    # Store data
    all_train_steps, all_train_values = [], []
    all_val_steps, all_val_values = [], []

    # Iterate through files (oldest to newest)
    for ef in event_files:
        try:
            ea = EventAccumulator(ef)
            ea.Reload()
            tags = ea.Tags().get("scalars", [])

            # Identify tags (flexible matching)
            t_tag = next((t for t in tags if "train" in t.lower() and ("loss" in t.lower() or "total" in t.lower())), None)
            v_tag = next((t for t in tags if ("valid" in t.lower() or "val" in t.lower()) and ("loss" in t.lower() or "total" in t.lower())), None)

            if t_tag:
                events = ea.Scalars(t_tag)
                for e in events:
                    all_train_steps.append(e.step)
                    all_train_values.append(e.value)

            if v_tag:
                events = ea.Scalars(v_tag)
                for e in events:
                    all_val_steps.append(e.step)
                    all_val_values.append(e.value)

        except Exception as e:
            print(f"⚠️ Error reading {os.path.basename(ef)}: {e}")

    # 3. Plot
    if all_train_steps or all_val_steps:
        plt.figure(figsize=(12, 6))

        if all_train_steps:
            t_data = sorted(zip(all_train_steps, all_train_values))
            plt.plot([x[0] for x in t_data], [x[1] for x in t_data], label="Train Loss", alpha=0.7)

        if all_val_steps:
            v_data = sorted(zip(all_val_steps, all_val_values))
            plt.plot([x[0] for x in v_data], [x[1] for x in v_data], label="Validation Loss", linewidth=2)

        plt.xlabel("Step / Epoch")
        plt.ylabel("Loss")
        plt.title("Training History (SepReformer)")
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()
    else:
        print("❌ Found event files but could not extract compatible 'loss' scalars.")
        if event_files:
             try:
                 ea = EventAccumulator(event_files[-1])
                 ea.Reload()
                 print(f"Available scalar tags in last file: {ea.Tags().get('scalars', [])}")
             except: pass
```

```python
import os
import textwrap
import torch
import yaml
import random

# --- Step 1: Re-Apply AttentionSpeakerSplit Patch ---
module_path = 'models/SepReformer_Base_WSJ0/modules/module.py'
NEW_CLASS_CODE = textwrap.dedent("""
class AttentionSpeakerSplit(torch.nn.Module):
    def __init__(self, in_channels: int, num_spks: int,
                 num_heads: int = 8, dropout: float = 0.05):
        super().__init__()
        self.in_channels = in_channels
        self.num_spks    = num_spks
        self.speaker_queries = torch.nn.Parameter(torch.empty(num_spks, in_channels))
        torch.nn.init.trunc_normal_(self.speaker_queries, std=0.02)
        self.cross_attn = torch.nn.MultiheadAttention(
            embed_dim=in_channels, num_heads=num_heads, dropout=dropout, batch_first=False
        )
        self.attn_norm = torch.nn.LayerNorm(in_channels)
        self.ff_gate = torch.nn.Sequential(
            torch.nn.LayerNorm(in_channels),
            torch.nn.Linear(in_channels, in_channels * 2, bias=True),
            torch.nn.GLU(dim=-1),
        )
        self.ff_norm = torch.nn.LayerNorm(in_channels)
        self.res_proj = torch.nn.Linear(in_channels, in_channels, bias=False)
        self.out_norm = torch.nn.GroupNorm(1, in_channels)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        B, D, T = x.shape
        kv = x.permute(2, 0, 1)
        q  = self.speaker_queries.unsqueeze(1).expand(-1, B, -1)
        attn_out, _ = self.cross_attn(q, kv, kv, need_weights=False)
        attn_out = self.attn_norm(attn_out + q)
        x_exp   = x.unsqueeze(1).expand(-1, self.num_spks, -1, -1)
        spk_ctx = attn_out.permute(1, 0, 2).unsqueeze(-1)
        res = self.res_proj(x.permute(0, 2, 1))
        res = res.permute(0, 2, 1).unsqueeze(1)
        out = x_exp * torch.sigmoid(spk_ctx) + res
        out_flat = out.reshape(B * self.num_spks, D, T).permute(0, 2, 1)
        ff_out   = self.ff_gate(out_flat)
        out_flat = self.ff_norm(out_flat + ff_out)
        out      = out_flat.permute(0, 2, 1).reshape(B, self.num_spks, D, T)
        out = self.out_norm(out.reshape(B * self.num_spks, D, T))
        return out
""")

# Patch module.py
if os.path.exists(module_path):
    with open(module_path, 'r') as f:
        src = f.read()
    if "AttentionSpeakerSplit" not in src:
        src = src.replace("class SpkSplitStage", "class _OldSpkSplitStage")
        src = src.replace("SpkSplitStage(", "AttentionSpeakerSplit(")
        src += NEW_CLASS_CODE
        with open(module_path, 'w') as f:
            f.write(src)
        print("✅ module.py re-patched.")
    else:
        print("ℹ️ module.py already patched.")

# Patch model.py
for fpath in ['models/SepReformer_Base_WSJ0/model.py']:
    if os.path.exists(fpath):
        with open(fpath, 'r') as f:
            src = f.read()
        if "SpkSplitStage(" in src:
            src = src.replace("SpkSplitStage(", "AttentionSpeakerSplit(")
            with open(fpath, 'w') as f:
                f.write(src)
            print(f"✅ {os.path.basename(fpath)} patched.")

# --- Step 2: Configure Training (Mini-Epochs & Mini-Validation) ---
config_path = 'models/SepReformer_Base_WSJ0/configs.yaml'
scp_dir = 'data/custom'
new_max_epoch = 200
target_batch_size = 12

train_subset_size = 1000  # 1000 samples for Training Epoch
val_subset_size = 500     # 500 samples for Validation (Faster)

print(f"✂️  Preparing subsets: Train={train_subset_size}, Val={val_subset_size}...")

def create_subset(files_map, size, prefix_replace_old, prefix_replace_new):
    new_map = {}
    if not os.path.exists(scp_dir): return {}

    try:
        # Read mixture file to check counts
        mix_file = files_map['mixture']
        with open(os.path.join(scp_dir, mix_file), 'r') as f:
            all_lines = f.readlines()

        num_samples = len(all_lines)
        if num_samples > size:
            indices = sorted(random.sample(range(num_samples), size))
            for key, fname in files_map.items():
                with open(os.path.join(scp_dir, fname), 'r') as f:
                    lines = f.readlines()
                subset_lines = [lines[i] for i in indices]

                new_fname = fname.replace(prefix_replace_old, prefix_replace_new)
                with open(os.path.join(scp_dir, new_fname), 'w') as f:
                    f.writelines(subset_lines)
                new_map[key] = new_fname
            print(f"   Created {prefix_replace_new}* files ({size} samples)")
            return new_map
        else:
            print(f"   Using full {prefix_replace_old}* set ({num_samples} <= {size})")
            return files_map
    except Exception as e:
        print(f"⚠️ Error creating subset for {prefix_replace_old}: {e}")
        return files_map

# 1. Create Train Subset
train_files = {'mixture': 'train_mix_clean.scp', 'spk1': 'train_s1.scp', 'spk2': 'train_s2.scp'}
mini_train_map = create_subset(train_files, train_subset_size, "train_", "train_mini_")

# 2. Create Validation Subset
val_files = {'mixture': 'val_mix_clean.scp', 'spk1': 'val_s1.scp', 'spk2': 'val_s2.scp'}
mini_val_map = create_subset(val_files, val_subset_size, "val_", "val_mini_")

# 3. Update Config
with open(config_path, 'r') as f:
    config = yaml.safe_load(f)

if 'config' in config:
    # Update Train Paths
    if mini_train_map:
        config['config']['dataset']['train']['mixture'] = mini_train_map.get('mixture')
        config['config']['dataset']['train']['spk1'] = mini_train_map.get('spk1')
        config['config']['dataset']['train']['spk2'] = mini_train_map.get('spk2')

    # Update Valid Paths
    if mini_val_map:
        config['config']['dataset']['valid']['mixture'] = mini_val_map.get('mixture')
        config['config']['dataset']['valid']['spk1'] = mini_val_map.get('spk1')
        config['config']['dataset']['valid']['spk2'] = mini_val_map.get('spk2')

    # Update Epochs
    config['config']['engine']['max_epoch'] = new_max_epoch
    config['config']['engine']['test_epochs'] = [new_max_epoch]

    # Update Batch Size
    current_bs = config['config']['dataloader'].get('batch_size', 12)
    if current_bs != target_batch_size:
        print(f"📉 Reducing batch_size from {current_bs} to {target_batch_size}.")
        config['config']['dataloader']['batch_size'] = target_batch_size
    else:
        print(f"✅ Batch size is already {target_batch_size}.")

with open(config_path, 'w') as f:
    yaml.safe_dump(config, f)

# --- Step 3: Start Training ---
print("\n📊 Expected Batches:")
if 'config' in config and 'dataset' in config['config']:
    train_scp = config['config']['dataset']['train']['mixture']
    val_scp = config['config']['dataset']['valid']['mixture']
    bs = config['config']['dataloader']['batch_size']

    def count_lines(fname):
        p = os.path.join(scp_dir, fname)
        if os.path.exists(p):
            with open(p, 'r') as f: return len(f.readlines())
        return 0

    n_train = count_lines(train_scp)
    n_val = count_lines(val_scp)

    print(f"  • Training Set:   {n_train} files / {bs} batch_size ≈ {n_train // bs} batches")
    print(f"  • Validation Set: {n_val} files / {bs} batch_size ≈ {n_val // bs} batches")

print("\n🚀 Resuming training...")
torch.cuda.empty_cache()
!TF_CPP_MIN_LOG_LEVEL=3 LOGURU_LEVEL=INFO PYTORCH_ALLOC_CONF=expandable_segments:True python run.py --model SepReformer_Base_WSJ0 --engine-mode train
```
